```python
import pandas as pd

df = pd.read_excel("BD/06_MATRICULAS_ED_SUPERIOR_VALPARAISO_2021.xlsx")

df
```

```python
edad_institucion = (
    df.groupby("TIPO DE INSTITUCION")["EDAD"]
    .agg(["count", "mean", "median", "std"])
    .round(1)
    .sort_values("mean")
)
print(edad_institucion)
```

El análisis muestra que la edad influye significativamente en el tipo de institución a la que ingresan los estudiantes.

Los estudiantes más jóvenes (alrededor de 22–23 años) se concentran principalmente en universidades, mientras que los estudiantes de mayor edad (promedio 27 años) se matriculan con más frecuencia en centros técnicos e institutos profesionales.

La desviación estándar creciente entre tipos de institución refleja una mayor diversidad etaria en los establecimientos técnicos, probablemente asociada a programas más flexibles y orientados a la reinserción educativa o laboral.
